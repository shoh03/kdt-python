## 비트 연산자
- `&`: bit 곱(and)
- `|`: bit 합(or)
- `^`: bit 배타적 논리합
- `~`: bit 뒤집기
- `>>`: 우측 쉬프트
- `<<`: 좌측 쉬프트

```python
# 연산을 처리할 때 최소 1byte(8bit) 단위로 처리함.
```

```python
# 12 & 11 -> 12는 이진수로 1100, 11은 1011, 1100과 1011을 &로 연결하면 같은 부분만 1, 나머지 0 -> 1000이 나옴
# 12 | 11 -> 1100과 1011의 각 자리수 중 하나라도 1이면 1, 둘 다 0이면 0 -> 1111이 나옴
# 12 >> 2 -> 1100을 화살표 방향으로 2칸 이동 = 이진수 11 = 3
```

```python
12>>2
```

## 진수 표현법
- bin() : 2진수
- hex() : 16진수
- int() : 10진수로 변경

```python
print(bin(10))
# 0b는 2진수라는 뜻
print(int(0b1010))
```

```python
print(hex(10)) 
# 0x는 16진수라는 뜻
# 16진수는 10이 16이기 때문에 9 다음에 10이 아니라 a가 옴. 8,9,a,b,c,d,e,f,10 순서
```

```python
print(bin(10))
print(bin(11))
```

```python
# 논리합
# 하나라도 1이면 1
print(10|11)
# 10진수 11 -> 2진수 1011
```

```python
# 논리곱 &
# 둘 다 1이면 1, 하나라도 0이면 0
print(10&11)
#10진수 10 -> 2진수 1010
```

```python
# xor(^)
# 모두 같으면 0, 아니면 1
print(10 ^ 11)
# 이진수 1011과 1010은 1,0,1이 같으니 셋 다 0, 마지막만 1과 0으로 다르니 1 -> 0001
```

```python
# not(~)
# ~해당값 => -(1 + 해당 값)
print(~10)
```

## 삼항 연산자(if else)
- 조건식이 True일 때 값 if 조건식 else 조건식이 false일 때 값
- `조건식 True일 때 수행` if `조건식` else `조건식 False일 때 수행`

```python
num = 10
'짝수' if num % 2 == 0 else '홀수'
```

```python
num1 = 23
num2 = 42

print(num2) if num1 < num2 else print(num1)
```
