```python
import requests
from bs4 import BeautifulSoup as bs
import json

aladin_url = "https://www.aladin.co.kr/shop/common/wbest.aspx?BranchType=1"
headers = {"User-Agent":"Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/152.0.0.0 Safari/537.36"}

response = requests.get(aladin_url, headers=headers)

aladin_soup = bs(response.text, "html.parser")

#제목
titles = aladin_soup.select(".ss_book_list .bo3")
title_list = list(map(lambda title: title.text, titles))
print(title_list)

#가격
#원가
tag_list = aladin_soup.select(".ss_book_box table table .ss_book_list:first-child li")
cost_list = [li.text.split("→")[0].strip().replace("원","") for li in tag_list[3::5]]
print(cost_list)

#할인가
prices = aladin_soup.select(".ss_book_list .ss_p2")
price_list = list(map(lambda price: price.text, prices))
print(price_list)

#별점
#별점이 없는 책들도 있어서 len() 사용했을 때 길이가 다르게 나오는 것 같습니다!
stars = aladin_soup.select(".ss_book_list .star_score")
star_list = list(map(lambda star: star.text, stars))
print(star_list)
```
