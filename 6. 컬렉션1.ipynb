# 컬렉션(list, tuple, set, dict)
1. list []
    - 추가, 수정, 삭제 언제나 가능
    - [] 대괄호로 묶어주고 ,(콤마)로 값을 구분한다

2. tuple ()
   - 한 번 생성되고나면 변경이 불가능
   - () 소괄호로 묶어주고 ,(콤마)로 값을 구분한다

3. set {}
   - 순서가 없는 자료 구조(비시퀀스)
   - 중복된 값을 저장이 불가능
   - {} 중괄호로 묶어주고 ,(콤마)로 값을 구분한다

4. dict {key:value}
   - 키 + 값이 한 쌍으로 관리하며 ,(콤마)로 값을 구분한다
   - 키는 반드시 문자열이어야 한다
   - key는 중복 값을 허용하지 않으며, value는 중복값을 허용한다
   - key를 통해 value를 가져온다

## ☆딕셔너리(dict)

```python
# 선언
di1 = {}
di2 = dict()
print(type(di1), type(di2))

# 빈 set -> {} 아니고 set()
```

```python
di3 = {
    'a':'apple',
    'b':'banana',
    'c':'cherry',
}

print(di3)
print(di3['a'])
```

```python
# 값을 조회
# 1 대괄호 접근법
print(di3['a'])

# 2 get
print(di3.get('a'))
```

```python
# d를 조회
# print(di3['d'])
# -> 대괄호 접근법은 없는 키에 접근했을 때 오류 발생
print(di3.get('d','대체값'))
# -> get으로 접근하면 없는 키에 접근했을 때 대체값 리턴, 대체값 지정 안했으면 None 리턴
```

```python
di3['a'] = 'alpha'

print(di3)
```

```python
di3 = {
    'a':'apple',
    'b':'banana',
    'c':'cherry',
    # 수정
    'a':'alpha',
}

print(di3)
```

```python
# .(dot) 접근법
print(list(di3.keys()))
print(list(di3.values()))
print(list(di3.items()))
```
