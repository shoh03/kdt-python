```python
# [문제 3] 중급: 상속과 다형성 (Employee Hierarchy)
# 난이도: ★★★☆☆  
# 핵심 개념: 클래스 상속(Inheritance), 메서드 오버라이딩(Overriding), super()
#
# [문제 설명]
# 직원 정보를 관리하는 클래스 구조를 설계하세요.
#
# 1. 부모 클래스 Employee:
#    - 이름(name)과 기본 월급(base_salary)을 속성으로 가집니다.
#    - get_salary() 메서드는 기본 월급(base_salary)을 반환합니다.
#
# 2. 자식 클래스 Developer:
#    - Employee를 상속받으며, 추가로 보너스(bonus) 속성을 가집니다.
#    - get_salary()를 오버라이딩하여 base_salary + bonus를 반환합니다.
#
# 3. 자식 클래스 Salesperson:
#    - Employee를 상속받으며, 실적(sales_count) 속성을 가집니다.
#    - get_salary()를 오버라이딩하여 base_salary + (sales_count * 50000)을 반환합니다.
```

```python
class Employee:
    def __init__(self, name, base_salary):
        self.name = name
        self.base_salary = base_salary

    def get_salary(self):
        return self.base_salary

class Developer(Employee):
    def __init__(self, name, base_salary, bonus):
        super().__init__(name, base_salary)
        self.bonus = bonus

    def get_salary(self):
        return self.base_salary + self.bonus

class Salesperson(Employee):
    def __init__(self, name, base_salary, sales_count):
        super().__init__(name, base_salary)
        self.sales_count = sales_count

    def get_salary(self):
        return self.base_salary + (self.sales_count * 50000)
```
