```python
musinsa_soup = bs(chrome.page_source, "html.parser")
```

```python

snap_images = [i.select_one("img")["src"] for i in musinsa_soup.select(".SnapFeedCard__Container-sc-e43ff597-0")]
snap_images
```

```python
import os
base_path = r"C:\kdt_0900_osh\python\resource"
```

```python
folder_name = r"musinsa"

musinsa_path = os.path.join(base_path, folder_name)
# ↓ 폴더 없을때만 생성
os.makedirs(musinsa_path, exist_ok = True)
```

```python
for image_url in snap_images:
    file_name = image_url.split("/")[-1].split("?")[0]
    file_path = os.path.join(musinsa_path, file_name)
    
    response = requests.get(image_url, stream=True)

    with open(file_path, "wb") as f:
        for chunk in response.iter_content(chunk_size=8192):
            f.write(chunk)

print("파일 저장 완료")
```

```python
hope_soup = bs(chrome.page_source, "html.parser")
```

```python
stars = hope_soup.select(".lego_movie_pure_star .area_text_box")

star_list = [float(i.text.replace("별점(10점 만점 중)","")) for i in stars]
hoburo_list = []

for star in star_list:
    if star == 5:
        hoburo_list.append("중립")
    elif star > 5:
        hoburo_list.append("호")
    else:
        hoburo_list.append("불호")

final_star_list = list(zip(star_list, hoburo_list))
len(final_star_list[0:-6])
```
