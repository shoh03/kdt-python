```python
def safe_calc(original_function):
    def exception_wrapper(*args):

        try:
            
            return original_function(*args)
        except ZeroDivisionError as e:
            print('0으로 나눌 수 없습니다.')
        except ValueError as e:
            print('정수만 입력')
        except TypeError as e:
            print('정수만 계산할 수 있습니다.')
        except Exception as e:
            print('알 수 없는 오류')
    return exception_wrapper
```

```python
@safe_calc
def device(num1,num2):
    return int(num1) /int(num2)
```

```python
device(1,0)
```

```python
@safe_calc
def add(num1,num2):
    return num1 + num2
```

```python
add('1',2)
```

```python
# keywards 같이 쓰기
# 유저의 dict를 넘기면, 유저 정보를 모두 출력
```

```python
def info_user(**keywards):
    print(f'{keywards['name']}\n {keywards['age']}\n{keywards['address']}')
```

```python
info_user(name='홍길동')
```

```python
def safe_value(original_function):
    def exception_wrapper(*args,**keywards):
        try:
            print('args',args)
            print('keywards', keywards)
            return original_function(*args, **keywards)
        except KeyError as e:
            print(f'{e}를 입력하세요')
        except Exception as e:
            print('알 수 없는 오류')
            print(e)
    return exception_wrapper
```

```python

@safe_value
def info_user(**keywards):
    print(f'{keywards['name']}\n {keywards['age']}\n{keywards['address']}')
```

```python
info_user(name='홍길동')
```
