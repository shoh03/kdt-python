```python
fruit_list = []
while True:
    fruit_name = input('과일의 이름을 입력하세요. 추가된 값을 지우려면 지우고 싶은 값을 작성하고 앞에 back을 붙이세요. 입력을 종료하려면 exit를 입력하세요')
    if fruit_name[0:4] == 'back' and fruit_name[4:] in fruit_list:
        fruit_list.remove(fruit_name[4:])
        continue
    elif fruit_name[0:4] == 'back' and fruit_name[4:] not in fruit_list:
        print('입력된 적 없는 과일 이름입니다.')
        continue
    if fruit_name == 'exit':
        break
    fruit_list.append(fruit_name)

print(fruit_list)

# stack 자료구조 : 먼저 들어온 값이 나중에 나가는 구조
```

```python
math_scores = [90, 85, 50, 70]
science_scores = [90, 34, 78, 45, 3, 4, 5]
english_scores = [89, 78, 56, 45, 4, 6]
names = ['승현','지혁','길동','지호','위위']

# 성적이 가장 우수한 학생 순서대로 순위 매기기
li1 = list(zip(math_scores, science_scores, english_scores, names))
li2 = list(map(lambda data: (data[0], data[1], data[2], data[3], data[0] + data[1] + data[2]), li1))
li3 = sorted(li2, key=lambda student: student[4], reverse = True)

for idx, student in list(enumerate(li3)):
    print(f'{idx + 1}위 {student[3]}님, 총 점: {student[4]}점')
```
