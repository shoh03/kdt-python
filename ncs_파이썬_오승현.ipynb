```python
chat = input('채팅을 입력하세요: ')

remove_list = ['바보','모지리','인공저능아']

chat_list = chat.split()

print(chat_list)

for i in remove_list:
    for j in chat_list:
        if i in j:
            chat_list[chat_list.index(j)] = j.replace(i,'****')

print(' '.join(chat_list))
```

```python
# 5.
# 마켓(Market) 클래스
# 1) 마켓은 이름(name), 판매 상품(product), 수입(income)의 필드를 갖는다.
# 2) 판매 상품은 다음과 같다
# - 이름(name)은 “토마토”이며, 토마토의 가격(price)은 3000원, 재고(stock)은 10개이다.
# 3) 마켓은 유저에게 상품을 판매할 수 있는 함수(sell)을 가지고 있다.
# - 유저가 가지고 있는 돈이 상품의 판매 가격보다 적다면 판매할 수 없다.
# - 유저가 구매하는 개수보다 재고의 개수가 적다면 판매할 수 없다.
# - 상품 판매시 유저가 가지고 있는 할인율이 적용되어 판매가 된다.

# ex) 유저의 할인율이 10%라면 2700원이 결제된다.

# - 판매된 상품의 모든 수익금은 income에 누적된다.

 

# 구매자(Buyer) 클래스
# 1) 구매자는 이름(name), 돈(money), 할인율(discount)의 필드를 갖는다.

# 2) 총 3명의 Buyer를 각각 객체화
# - 홍길동, 15000, 5
# - 장보고, 10000, 20
# - 이순신, 30000, 50
 

# 구매자의 할인율을 적용받아 마켓에서 토마토를 구매를 아래와 같이 구매하였다.

# 마켓의 총 수익과 각 구매자의 잔액을 출력하시오.

# - 홍길동 6개 구매

# - 장보고 2개 구매

# - 이순신 8개 구매
```

```python
class Market:
    def __init__(self, name, product, income = 0):
        self.name = name
        self.income = income
        self.product = product

    def sell(self, buyer, count):
        discount_price = int(self.product['price'] * (1 - buyer.discount / 100))
        if count > self.product['stock']:
            print(f'잔여 상품이 부족합니다. 잔여 수량 {self.product['stock']}')
            return
            
        if discount_price * count > buyer.money:
            print(f'''{buyer.name}님의 잔액이 부족합니다. 
        잔액 : {buyer.money}원
        ''')
            return

        
        buyer.money -= discount_price * count
        self.income += discount_price * count
        self.product['stock'] -= count
        print(f'''{buyer.name}님, {self.product['name']} {count}개 구매완료. 
        최종가격 : {discount_price}원
        지불액 : {discount_price * count}원
        잔액 : {buyer.money}원
        현재 총 수익 : {self.income}원
        ''')
    

class Buyer:
    def __init__(self, name, money, discount):
        self.name = name
        self.money = money
        self.discount = discount


karrot = Market('당근마켓',{'name':'토마토','price':3000,'stock':10})
bungae = Market('번개장터',{'name':'감자','price':1100,'stock':25})


hong = Buyer('홍길동', 15000, 5)
jang = Buyer('장보고', 10000, 20)
lee = Buyer('이순신', 30000, 50)
```

```python
karrot.sell(hong,6)
karrot.sell(jang,2)
karrot.sell(lee,8)
```
