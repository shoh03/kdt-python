## 지역변수
   - 함수 내부에서 선언된 변수이며, 함수에서만 사용이 가능하다.

## 전역변수
   - 함수 밖에서 선언한 변수, 전역에서 사용이 가능하다.

## global문
   - global문을 사용하면 함수 안에서도 전역변수의 값을 수정이 가능하다

영역의 구분은 들여쓰기를 기준으로 함

```python
# 전역 변수
value = 100

# 지역 변수
def change_value():
    value = 20

change_value()
print(value)

def change_value():
    # 이렇게 하면 전역 변수 호출
    global value
    value = 20

change_value()
print(value)
```
