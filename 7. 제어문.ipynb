```python
num1 = 21
num2 = 20

if num1 > num2:
    print(f'큰 수 = {num1}, 작은 수 = {num2}')
elif num2 > num1:
    print(f'큰 수 = {num2}, 작은 수 = {num1}')
else:
    print('두 수는 같습니다')
```

```python
name = '오승현'
math_score = 23
eng_score = 44
kor_score = 99
total_score = math_score + eng_score + kor_score
aver_score = (total_score)/3

if aver_score >= 90:
    grade = 'A'
elif aver_score >= 80:
    grade = 'B'
elif aver_score >= 70:
    grade = 'C'
elif aver_score >= 60:
    grade = 'D'
else:
    grade = '과락'

print(f'{name}님의 총 점은 {total_score}점, 평균은 {aver_score:.2f} 점으로 {grade} 등급입니다.')
# round(값,2) : 소수점 3자리수에서 반올림
round(aver_score, 2)
```
