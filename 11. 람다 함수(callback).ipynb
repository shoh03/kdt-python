```python
# 두 수를 더해서 출력
def sum1(num1, num2, callback):
    callback(num1+num2)

sum1(10,20,print)
```

```python
# 성과 이름을 전달하면 풀네임을 만들고 님을 붙여서 반환

def get_fullname(last_name, first_name, callback):
    callback(last_name + first_name)

def add_nim(name):
    return name + '님'

get_fullname('홍','길동',add_nim)

# -> 이렇게 하면 add_nim에는 return 값이 존재하지만 그 return 값은 최종적으로 반환되는 것이 아니라
# get_fullname 내에서 반환되므로 get_fullname 내에 return값이 정의되어있지 않다면 최종적인 return 값은 없음.
# 즉, 위의 함수는 
# def get_fullname(last_name, first_name, callback):
#    '홍길동님'
# 과 같은 형태인 것.

def get_fullname(last_name, first_name, callback):
    return callback(last_name + first_name)

def add_nim(name):
    return name + '님'

get_fullname('홍','길동',add_nim)

# 이렇게 해야 
# def get_fullname(last_name, first_name, callback):
#    return '홍길동님'
# 과 같이 작동하여
# 최종적인 return 값이 존재함.
```

```python
def five_name(name,callback):
    for i in range(5):
        callback(name)

five_name('오승현',print)

def print_name(name):
    print(name)

def repeat(name, callback):
    for i in range(5):
        callback(name)

repeat('오승현',print_name)

repeat('홍길동', lambda name: print(name))
```

```python
# 두 수를 더하고 2배를 곱한 결과에 10배를 만들고 반환

def add(num1,num2,callback):
    return callback(num1 + num2)

def get_square(num, callback):
    return callback(num ** 2)

def get_mul_10(num):
    return num * 10

# add(10,20,get_square)
# -> 이렇게 하면 get_square의 callback함수 자리에는 함수를 넣어줄 수 없음.

add(10, 20, lambda add_result: get_square(add_result, get_mul_10))
# 이렇게 하면 가능
add(10, 20, lambda add_result: get_square(add_result, lambda result: result * 10))
# 이렇게 해도 가능
```

## 람다 실습

```python
# 두 정수의 결과를 곱하고 출력
# 횟수를 전달하면 횟수까지의 합을 반환
# 1~100까지 짝수만 더한 뒤 결과 값에 2를 곱한 후 출력
# 두 수를 더하고 홀수인지 짝수인지 비교 후 출력
```

```python
def multiple(num1,num2,callback):
    callback(num1 * num2)

multiple(3,4,lambda x: print(x))
```

```python
def add(num, callback):
    hap = 0
    for i in range(1,num+1):
        hap = hap + i
    return callback(hap)

add(5, lambda x: x)

def get_total(count, callback):
    return callback(count)

def repeat(count):
    total = 0
    for i in range(count):
        total += i + 1
    return total

get_total(10, repeat)

# 출력 + 추가 함수
def get_total(count, callback):
    callback(count)

def repeat(count, callback):
    total = 0
    for i in range(count):
        total += i + 1
    callback(total)

get_total(10, lambda total: print(total))
```

```python
def add_hund(hund, callback):
    num = 0
    for i in range(1,hund + 1):
        if i % 2 == 0:
            num = num + i

    return callback(num)

add_hund(100, lambda x: x * 2)

def get_total_even(callback):
    callback(sum([i for i in range(0, 101, 2)]))

def multiple(num):
    print(num * 2)

get_total_even(multiple)
```

```python
def hol_chak(num1,num2,callback):
    return callback(num1 + num2)

hol_chak(3,2,lambda x: '짝' if x % 2 == 0 else '홀')

def add(num1,num2, callback):
    callback(num1 + num2)

def confirm_odd_even(num, callback):
    callback('짝수' if num % 2 == 0 else '홀수')

add(10,20, lambda x: confirm_odd_even(x, print))
```

## filter, map, reduce

### 1: map 함수
- iterable 한 객체들을 순차적으로 순회하여 결과를 변환하고 반환하는 함수

```python
li1 = [(i + 1) * 10 for i in range(10)]
li1
```

```python
for idx, value in enumerate(li1):
    li1[idx] = value * 10

print(li1)

# -> 원본이 바뀜
```

```python
li2 = [(i + 1) * 10 for i in range(10)]
li2
```

```python
list(map(lambda num: num * 10,li2))
```

```python
# li3 -> 홀수면 true 짝수면 false
li3 = [i + 1 for i in range(50)]
print(li3)
```

```python
li4 = list(map(lambda num: num % 2 == 1, li3))
print(li3)
print(li4)
# 원본이 바뀌지 않음.
```

### 2. filter 함수
- iterable한 객체들을 순차적으로 순회하여 조건식을 만족하는 값들만 모아서 반환해주는 함수

```python
# filter
# 짝수만 걸러내기
# filter(func,a)에서 func를 만족하는 원소만 따로 모아서 반환.
print(list(filter(lambda num: num % 2 == 0, li3)))
```

```python
li5 = [chr(ord('a') + i) for i in range(26)]
print(li5)
```

```python
# acegik....(짝수 인덱스들의 알파벳만 걸러내기)
li6 = list(filter(lambda tu: tu[0] % 2 ==0 , enumerate(li5)))
print(li6)
```

```python
li7 = list(map(lambda c: c[1], li6))
print(li7)
```

### 3. reduce 함수
- iterable한 객체들을 순차적을 순회하여 모든 값들을 누적시켜서 반환해주는 함수

```python
# 1~10까지 모든 합을 구하기
from functools import reduce

li11 = [i + 1 for i in range(10)]
li11

reduce(lambda acc,crr: acc + crr, li11)
```

```python
# 1~10까지 리스트에 담고 값이 5보다 크면 5배로 변경
# 1~ 30까지 배열에 담고 3을 곱한 뒤 출력
# 소문자 알파벳들을 담고 아래와 같이 변경 후 출력
# aBcD...
```

```python
lis = [i + 1 for i in range(10)]
list(map(lambda x: x * 5 if x > 5 else x, lis))
```

```python
lislis = [i+1 for i in range(30)]
print(list(map(lambda x: x * 3, lislis)))
```

```python
small = [chr(i) for i in range(ord('a'),ord('z') + 1)]
large = [chr(i) for i in range(ord('A'),ord('Z') + 1)]
idx_dict = {}
for i,k in enumerate(small):
    idx_dict[k] = i

print(list(map(lambda x: x if idx_dict[x] % 2 == 0 else large[idx_dict[x]], small)))
```
