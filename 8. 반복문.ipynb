```python
# 마피아 게임에서 사람들에게 마피아 2명, 의사 1명, 경찰 1명, 나머지는 시민 직업을 부여하는 코드를 작성하시오.(사람 이름과 수는 입력받은대로, 다만 최소 인원수는 4명)
par_list = []
job_list = ['마피아','마피아','의사','경찰']
par_job = {}
while True:
    par = input("최소 4명 이상의 이름을 입력하세요. 모두 입력하셨다면 '시작'을 입력하세요:")
    if par in par_list:
        print('같은 사람의 이름을 2번 입력하셨습니다')
    if par != '시작':
        par_list.append(par)
    else:
        if len(par_list) < 4:
            print('최소 인원수는 4명입니다. 추가적인 이름을 입력하세요.')
        else:
            break
for i in range(0,len(par_list)):
    job_list.append('시민')
for j in range(0,len(par_list)):
    par_job[par_list[j]] = job_list[j]

print(par_job)
```

```python
text = "abcccddeaddddadwefda"

most_char = text[0]
max_count = 1

current_char = text[0]
current_count = 1

for i in range(1, len(text)):
    if text[i] == current_char:
        current_count += 1
    else: 
        current_char = text[i]
        current_count = 1

    if current_count > max_count:
        most_char = current_char
        max_count = current_count

result = f"가장 연속된 글자가 많은 문자형: {most_char}\n연속 횟수: {max_count}번"
print(result)
```
