# 변수 선언 시 주의사항
1. 되도록 선언부(영역 상단)에 한꺼번에 선언한다
2. 초기화를 해준다

## 변수명 선언 시 주의사항
1. 문자로 시작해야 한다.
2. 특수 문자는 사용할 수 없다.(_ 제외)
3. 소문자로 시작한다
4. 공백을 사용할 수 없다
5. 명사로 사용한다
6. 되도록 한글은 사용하지 않는다.
7. 뜻이 있는 단어를 사용한다.  
        ex)
   - a, b, c, d, e ...(X)
   - data, number, age, name ...(O)

```python
letter = 'a'
tf = False
int_num = '5'
float_num = '10.7'

print(int(bool(letter)))
print(int(tf))
print(int(int_num))
print(int(float(float_num)))
```
