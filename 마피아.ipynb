```python
import random
import pandas as pd
import numpy as np
from IPython.display import display
import time
```

```python
def add_list(**keywards):
    li = []
    for i in keywards:
        li += [i] * keywards[i]
    return li
```

```python
# 참가자 받기

par_list = []
job_list = []

print('참가자 이름을 입력하세요. 참가자를 제거하려면 \'back 참가자\'와 같이 입력하세요. 게임을 시작하려면 start를 입력하세요. \n최소인원 : 8명, 최대 인원 12명')
while True:
    if (par := input('참가자 :')) == 'start':
        par_num = len(par_list)
        if par_num < 8:
            print(f'최소 인원을 충족하지 못했습니다. 인원을 {8-par_num}명 더 입력해주세요.')
            continue
        if par_num > 12:
            print(f'최대 인원을 초과했습니다. 인원을 {par_num - 12}명 빼주세요.')
            continue
        print(f'최종 참가자 명단: {par_list}')
        par_num = len(par_list)
        break
    elif 'back' in par:
        par_list.remove(par.replace('back','').replace(' ',''))
        continue
    if par in par_list:
        print('이미 추가된 이름입니다.')
        continue
    par_list.append(par)
    print(f'현재 참가자 명단 : {par_list}')
```

```python
# 직업 

print('''마피아 인원수, 경찰 인원수, 의사 인원수를 입력하세요. 기본 설정값으로 진행하려면 "default"를 입력하세요.
ex) 4,2,1 -> 마피아 4명, 경찰 2명, 의사 1명\n
기본설정값:''')
display(pd.DataFrame([
    [2, 1, 1],
    [3, 2, 1],
    ],columns=['마피아','경찰','의사'],
    index=['8~10명','11~12명']
                    )
        )

if (num_str := input('마피아, 경찰, 의사 인원수(순서대로 입력): ')) == 'default':
    if 8 <= len(par_list) <=10:
        job_list = add_list(mafia=2,police=1,doctor=1)
    else:
        job_list = add_list(mafia=3,police=2,doctor=1)
else:
    job_num_dict = dict(zip(['mafia','police','doctor'],map(lambda a: int(a),num_list.replace(' ','').split(','))))
    job_num_str = f'mafia = {job_num_dict['mafia']}, police = {job_num_dict['police']}, doctor = {job_num_dict['doctor']}'
    job_list = eval(f'add_list({job_num_str})')

random.shuffle(job_list)

par_job = dict(zip(par_list,job_list))
print(par_job)

```
