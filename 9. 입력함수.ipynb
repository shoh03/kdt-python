## 입력함수: input()
- 사용자에게 입력받아 문자열 값으로 반환하는 함수

```python
name = input('이름을 입력하세요\n')
```

```python
num1 = int(float(input('첫 번째 정수 입력:\n ex) 20\n')))
num2 = int(float(input('두 번째 정수 입력:\n ex) 20\n')))
result = f'두 수의 합은: {num1 + num2}'

print(result)
```

```python
dan = input('구구단수를 입력하세요 /nex) 3단 -> 3 :')
if float(dan) % 1 == 0:
    for i in range(1,10):
        print(f'{int(dan)} × {i} = {int(dan) * i}')
else:
    print(f'{dan}은 정수가 아닙니다. 정수를 입력해주세요.')
```
