```python
import math

# math.ceil = 올림
print(math.ceil(3.14))

# math.floor = 내림
print(math.floor(3.99))

# round() = 반올림
print(round(2.5))

# math.factorial = 팩토리얼
print(math.factorial(5))

# math.pi = 원주율(함수가 아니라 값임)
print(math.pi)
```

```python
import random

# .randint(이상, 이하)

lotto = []

for i in range(6):
    lotto.append(random.randint(1,45))

print(lotto)
```

```python
import random

# .seed() = random한 값을 고정시킴
random.seed(2026)

lotto = []

for i in range(6):
    lotto.append(random.randint(1,45))

print(lotto)
```

```python
# .randrange(초과, 미만) = .randint()와 달리 초과 미민
print(random.randrange(0,10))
```

```python
# .choice() = 시퀀스 자료구조 중 1개를 선택하여 반환

students = ["김희수",'조수민','박지호','혜진','이상아','김진휘','정우영','오승현','오슬기','오지혁','이예나','채진성']

print(random.choice(students))
```

```python
# random.sample() = 여러 개 선택해서 반환

students = ["김희수",'조수민','박지호','혜진','이상아','김진휘','정우영','오승현','오슬기','오지혁','이예나','채진성']

random.sample(students, 4)
```

```python
from collections import Counter

# Counter(객체) = 리스트 또는 시퀀스 객체에서 개수를 세는 객체


fruits = "apple banana apple banana apple melon".split()
fruit_count = Counter(fruits)

print(fruit_count)
# -> fruit_count의 값은 dict 모양이지만 dict 타입이 아님.
```

```python
text = "hello, world, python"
text_count = Counter(text)

print(text_count["o"])
```

```python
# key 없어도 오류 안남
fruit_count = Counter({'apple': 3, 'banana': 2, 'melon': 1})
print(fruit_count["a"])

# key 없으면 오류
print({'apple': 3, 'banana': 2, 'melon': 1}["a"])
```
