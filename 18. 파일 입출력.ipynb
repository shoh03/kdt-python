## with 문
- resource를 사용한 뒤 자동으로 반납(close)하여 안전하게 관리해주는 문법

```python
with open(file_path,'w',encoding='utf-8') as f:
    f.write('하이')
    f.write('요')

    # with 구문이기 때문에 f.close()가 자동으로 실행됨(finally 효과도 포함되어 자동 실행)
```

## 파일 수정

```python
foods = ['돈까스','제육볶음','김치찌개','순두부찌개','된장찌개','순대국','센베이']

file_name = 'food_list.txt'
file_path = os.path.join(base_path, file_name)

with open(file_path, 'w', encoding='utf-8') as f:
    f.writelines([food + '\n' for food in foods])
```

```python
# 1. 읽어오기
with open(file_path, 'r', encoding='utf-8') as f:
    foods = f.readlines()
    foods[6] = '삼겹살\n'

with open(file_path, 'w', encoding='utf-8') as f:
    f.writelines(foods)
    print(f'{file_name}파일 수정이 완료되었습니다.')
```

```python
# 돈까스, 된장찌개 삭제

with open(file_path, 'r', encoding='utf-8') as f:
    souls = f.readlines()
    print(souls)
    souls.remove('돈까스\n')
    souls.remove('된장찌개\n')

print(souls)

with open(file_path, 'w', encoding='utf-8') as f:
    f.writelines(souls)

print(souls)
```

```python
foods = ['돈까스','제육볶음','김치찌개','순두부찌개','된장찌개','순대국','센베이']

file_name = 'food_list.txt'
file_path = os.path.join(base_path, file_name)

with open(file_path, 'w', encoding='utf-8') as f:
    f.writelines([food + '\n' for food in foods])
```

```python
with open(file_path, 'r', encoding='utf-8') as f:
    souls = f.readlines()
    print(souls)
    souls.remove('돈까스\n')
    souls.remove('된장찌개\n')

print(souls)

with open(file_path, 'w', encoding='utf-8') as f:
    f.writelines(tuple(souls))

print(souls)
```

```python
with open(file_path, 'r', encoding='utf-8') as f:
    foods = f.readlines()
    for food in foods:
        if food == '돈까스\n' or food == '된장찌개\n':
            foods.remove(food)

foods
with open(file_path, 'w', encoding='utf-8') as f:
    f.writelines(foods)
    print(f'{file_name}이/가 수정되었습니다.')
```

```python
# CSV(Comma Separated Value)
import csv

file_name = '동물미용실.csv'
file_path = os.path.join(base_path, file_name)

with open(file_path, 'r', encoding='cp949') as f:
    reader = csv.reader(f)

    header = next(reader)
    for row in list(reader)[:5]:
        print(row)
```

```python
file_name = '할일목록.csv'
file_path = os.path.join(base_path, file_name)

with open(file_path, 'w', encoding='utf-8-sig', newline='') as f:
    writer = csv.writer(f)
    # 첫 줄(열) 제목
    writer.writerow(['날짜','할일 목록'])
    writer.writerows([('2026-08-28','퇴실'), ('2026-08-28','복습'), ('2026-08-28','잠자기')])

    print(f'{file_name}이 생성되었습니다.')
```
