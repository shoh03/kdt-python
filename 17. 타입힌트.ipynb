## pydantic
- 데이터 검증과 직렬화를 위한 파이선의 라이브러리로 데이터 파이프라인 구축에 필수적으로 사용된다.
- https://pydantic.dev/docs/validation/latest/api/pydantic/base_model/

```python
# 같은 데이터이고, 같은 key 값이어도 같은 타입으로 보기 힘듦.
# 클래스
user1 = {
    'name':'홍길동',
    'age':20,
    'address':'서울시 강남'
}

cat1 = {
    'name':'야옹이',
    'age':5,
    'address':'서울시 강남'
}

class Person:
    def __init__(self, name: str, age: int, address: str):
        self.name = name
        self.age = age
        self.address = address

class Animal:
    def __init__(self, name: str, age: int, address: str):
        self.name = name
        self.age = age
        self.address = address

my_cat = Animal(**cat1)
my_hgd = Person(**user1)
print(my_cat.name)
```

```python
# type 비교
my_cat = Animal(**cat1)
my_hgd = Person(**user1)

print(my_cat.name)

print(my_cat)
print(my_hgd)

print(isinstance(my_cat, Person))
print(isinstance(my_hgd, Person))
```

```python
from pydantic import BaseModel, Field
```

```python
class Person(BaseModel):
    name:str = Field(description='사람 이름', min_length=2, max_length=30)
    age:int = Field(description='사람 나이', ge=1, le=150)
    address:str = Field(description='사람 주소', default='알 수 없음')

print(Person(name='홍길동', age=10, address='서울'))  
print(Person(name='홍길동', age='10', address='서울'))  # age에 str이 왔지만 int로 형변환이 가능하므로 자동 형변환 -> 오류 안남
print(Person(name='홍길동', age='r', address='서울')) # age에 int로 변환할 수 없는 값이 왔으므로 오류남
```

```python
class Person(BaseModel):
    name:str = Field(description='사람 이름', min_length=2, max_length=30)
    # strict : 자동 형변환까지 막는 엄격한 타입검증
    age:int = Field(description='사람 나이', ge=1, le=150, strict=True)
    address:str = Field(description='사람 주소', default='알 수 없음')

print(Person(name='홍길동', age=10, address='서울'))
print(Person(name='홍길동', age='10', address='서울'))  # age의 strict값이 True이므로 자동 형변환 안됨 -> 오류 남
print(Person(name='홍길동', age='r', address='서울'))
```
