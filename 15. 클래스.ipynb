```python
# 카페(cafe) 클래스 생성
# 메뉴들(menus) 메뉴 소개 메서드(introduce)

class Cafe:
    # default parameter
    def __init__(self, name, menus = []):
        self.name = name
        self.menus = menus

    def introduce(self):
        for idx, menu in enumerate(self.menus):
            print(f'{self.name} 메뉴{idx+1}: {menu}')
```

```python
# 실습
# 유저(User) 클래스를 생성하고 아래 코드를 완성하시오
# 유저는 이름(name), 돈(money)를 가지고 있다. 돈은 모두 10000원씩 존재한다
# 유저는 출금(withdraw) 메서드를 통해 유저의 돈을 출금할 수 있다.
# 아래에 코드를 완성하시오.

class User:
    
    def __init__(self,name,money = 10000):
        self.name = name
        self.money = money

    def withdraw(self,spend):
        if spend > self.money:
            print(f'잔액이 부족합니다. 잔액:{format(self.money,'_')}원')
            return -1
        self.money -= spend
        print(f'{self.name}님 {spend}원 출금완료.\n잔액:{self.money}원')
        return spend
```

```python
class Banking:
    
    def __init__(self, name, bank_name, money = 10000, password=0000, try_num = 0):
        self.name = name
        self.money = money
        self.password = password
        self.try_num = try_num
        self.bank_name = bank_name
        if password == 0000:
            print('비밀번호를 변경해주세요')

    def withdraw(self, input_pw, spend):
        if self.password != input_pw:
            self.try_num += 1
            if self.try_num >= 5:
                print(f'비밀번호를 5회 이상 틀려 계좌가 잠겼습니다. 잠금을 해제하려면 모바일 앱을 이용하시거나 {self.bank_name} 지점을 방문해주세요.')
                return -1
            print(f'비밀번호를 {self.try_num}회 틀리셨습니다. 5회 이상 틀릴 시 계좌 잠김.')
            return -1
            
        if spend > self.money:
            print(f'잔액이 부족합니다. 잔액:{self.money}원')
            return -1
            
        self.money -= spend
        print(f'{self.name}님 {spend}원 출금완료.\n잔액:{self.money}원')
        return spend

    def lock_off(self):
        self.try_num = 0
        return '비밀번호 시도 횟수가 초기화되었습니다.'

    def change_password(self, old_password, new_password):
        if old_password != self.password:
            print('비밀번호가 틀렸습니다')
            return -1
            
        self.password = new_password
        return '비밀번호가 변경되었습니다'

            
```

```python
class Locker:
    def __init__(self,owner,password,items=[]):
        self.owner = owner
        self.password = password
        self.items = list(items)

    def Locker_info(self,pw):
        if pw != self.password:
            print('비밀번호 틀림')
            return -1
            
        return f'사물함 내부 : {self.items}'

    def verify_password(self, view_password):
        return self.password == veiw_password

    def add_items(self, pw, *args):
        if not self.verify_password(pw):
            print('비밀번호 틀림')
            return
        
        for i in args:
            self.items.append(i)
        print(f'사물함 내부 : {self.items}')

    def get_item(self, pw, wanted_item):
        if not self.verify_password(pw):
            print('비밀번호 틀림')
            return -1
            
        if wanted_item not in self.items:
            print('사물함에 없는 물건')
            return -1
            
        return(self.items.pop(self.items.index(wanted_item)))
```

```python
class Animal:
    def __init__(self,name,foot_num):
        self.name = name
        self.foot_num = foot_num

    def eat(self, *food):
        for i in food:
            print(f'{i} 냠냠')
        if len(food) < 3:
            print('아직 배고파')
        else:
            print('이제 배불러')


class Cat(Animal):
    def __init__(self, name):
        super().__init__(name, foot_num = 4)

    def crawl(self, crawl_num):
        print(f'앞 발로 {crawl_num}번 할퀴기!')
        if crawl_num < 4:
            return '효과는 미미했다...'
        return '효과가 굉장했다!'
```

## 마커(Maker)

```python
# 동물들
# 초식, 육식, 잡식


# 초식
class HerbivoreMaker:
    pass

# 육식
class CarnivoreMaker:
    pass

# -> 마커로 사용하기 위한 클레스들은 이름 뒤에 Maker을 붙이기(개발자끼리의 약속)

class Animal:
    def __init__(self, name):
        self.name = name

class Bear(Animal, CarnivoreMaker):
    def __init__(self,name):
        super().__init__(name)

class Cow(Animal, HerbivoreMaker):
    def __init__(self,name):
        super().__init__(name)

class Dog(Animal):
    def __init__(self,name):
        super().__init__(name)

class Tiger(Animal,CarnivoreMaker):
    def __init__(self,name):
        super().__init__(name)

class Cat(Animal):
    def __init__(self,name):
        super().__init__(name)
        
```

```python
animals = [Bear("곰"), Cow('소'), Dog('개'), Tiger('호랑이'), Cat('고양이')]
```

```python
for animal in animals:
    # 육식동물이라면
    if isinstance(animal, CarnivoreMaker):
        print(animal.name)
```

## 객체간의 프로그래밍

```python
# 은행,유저,송금,출금

class Bank:
    def __init__(self, name, bank_money = 1000000):
        self.name = name
        self.bank_money = bank_money

    # 입금
    # 비밀번호, 통장, 돈, 아이디
    def deposit(self, user, money):
        # 은행에 +만원
        self.bank_money += money
        # user 돈에 -만원
        user.user_money -= money
        print(user.user_money)
        

    # 출금
    # 돈, 카드비밀번호, 통장, 정보
    def withdraw(self, user, money):
        # 은행에서 -
        self.bank_money -= money
        # user에 +
        user.user_money += money

class User:
    def __init__(self, name, user_money = 0):
        self.name = name
        self.user_money = user_money

kb = Bank('국민은행')
jh = User('지효', 100000)

```

```python
kb.deposit(jh, 10000)
print(jh.user_money)
kb.withdraw(jh, 50000)
print(jh.user_money)
```

```python
# 쿠폰 할인률
# 상점(Store), 고객(Customer)
# 판매(sell)

class Store:
    def __init__(self, name):
        self.name = name
        self.product = {'name':'모자', 'price':10000}

    def sell(self, customer, count):
            price = self.product['price']
            discounted_price = price - int(price * (customer.discount / 100))
            customer.money -= price * count
            print(f'원래 상품 가격 : {price}원, 할인가 : {discounted_price}원')

class Customer:
    def __init__(self, name, money, discount):
        self.name = name
        self.money = money
        self.discount = discount
```

```python
store = Store('인터넷마켓')

hong = Customer('홍길동', 50000, 15)
jang = Customer('장보고', 100000, 80)

store.sell(hong, 1)
```

```python
# 마켓 문제
# - 마켓(Market) 클래스
# 1) 마켓은 이름(name), 판매 상품(product), 수입(income)의 필드를 갖는다.
# 2) 판매 상품은 다음과 같다
# - 이름(name)은 “토마토”이며, 토마토의 가격(price)은 3000원, 재고(stock)은 10개이다.
# 3) 마켓은 유저에게 상품을 판매할 수 있는 함수(sell)을 가지고 있다.
# - 유저가 가지고 있는 돈이 상품의 판매 가격보다 적다면 판매할 수 없다.
# - 유저가 구매하는 개수보다 재고의 개수가 적다면 판매할 수 없다.
# - 상품 판매시 유저가 가지고 있는 할인율이 적용되어 판매가 된다.

# ex) 유저의 할인율이 10%라면 2700원이 결제된다.
# - 판매된 상품의 모든 수익금은 income에 누적된다.


# - 구매자(Buyer) 클래스
# 1) 구매자는 이름(name), 돈(돈), 할인율(discount)의 필드를 갖는다.
# 2) 총 3명의 Buyer를 각각 객체화
# - 홍길동, 15000, 5
# - 장보고, 10000, 20
# - 이순신, 30000, 50
 
# 구매자의 할인율을 적용받아 마켓에서 토마토를 구매를 아래와 같이 구매하였다.
# 마켓의 총 수익과 각 구매자의 잔액을 출력하시오.
# - 홍길동 6개 구매
# - 장보고 2개 구매
# - 이순신 8개 구매
```

```python
class Market:
    def __init__(self, name, income=0):
        self.name = name
        self.product = {'name':'토마토', 'price':3000, 'stock':10}
        self.income = income
    
    def sell(self,buyer,num):
        discount_price = int(self.product['price'] * (1 - buyer.discount / 100) * num)
        if buyer.money < discount_price:
            print(f'잔액이 부족합니다. 남은 잔액 : {buyer.money}, 필요 금액 : {discount_price}')
            return
        if self.product['stock'] < num:
            print(f'구매하려는 상품의 수량이 부족합니다. 남은 수량 : {self.product['stock']}')
            return
            
        buyer.money -= discount_price
        self.product['stock'] -= num
        self.income += discount_price
        print(f'''상품이 판매되었습니다. 

        단일 제품 최종 가격 : {int(discount_price / num)}
        판매수익 : {discount_price}
        총수익 : {self.income}
        상품 잔여 수량 : {self.product['stock']}
        구매자 잔액 : {buyer.money}''')

    def show_total_income(self):
        print(f'총수익 : {self.income}\n상품 잔여 수량 : {self.product['stock']}')


class Buyer:
    def __init__(self, name, money, discount):
        self.name = name
        self.money = money
        self.discount = discount



karrot = Market('당근마켓')

hong = Buyer('홍길동',15000, 5)
jang = Buyer('장보고',10000, 20)
lee = Buyer('이순신',30000, 50)
```
