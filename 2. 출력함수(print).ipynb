# 제어문자
- 반드시 따옴표 안에서 작성.  
  문자열 안에서 특수한 기능을 가지는 문자들을 의미함.
- \n : new line, 개행문자(줄바꿈 문자)
- \t : tab, 위 아래 줄 간격 맞춰 띄기
- \" : "를 표현
- \' : '를 표현
- \\ : \를 표현

# 문자열('',"")
- 컴퓨터가 명령어가 아닌 문자열 값임을 구분할 수 있게 사용하는 특수한 기호
- 즉, 문자열은 문자들의 나열

```python
# 함수 : 이름 뒤에 소괄호 붙음, 어떤 기능 또는 값
print("\'오승현\', \n\'24세\'(만 22세)")
```

```python
print('오승현 \n취미\n1: 칵테일\n2: 노래\n3: 게임')
```

```python
print('홍길동',end='')
print('입니다')
```

## 규칙
- 1. 내용 안에 들어가는 내용 -> 안으로 들여쓰기([tab])해주는 게 규칙
  2. 내용들끼리는 같은 depth(깊이)의 들여쓰기를 사용해야 한다.
     만약 들여쓰기를 잘못 넣었다면 [Shift] + [tab]으로 바깥으로 꺼내준다.
  3. 띄어쓰기(' ') (== 공백문자)를 잘못 입력하면 depth로 인식할 수 있음.

```python
print("""
    안녕
    난
    오승현
""")
```

```python
print('100 * "20" * 20 = ?')
```
