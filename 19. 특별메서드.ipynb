```python
class User:
    name: str
    age: int
    phone: str

    def __init__(self, name, age, phone):
        self.name = name
        self.age = age
        self.phone = phone

    def __str__(self):   # print() 했을 때 나오는 값
        return f'User(name: {self.name}, age: {self.age}, phone:{self.phone})'
```

```python
print(User('홍길동', 20, '010-1234-1234'))
```

```python
class Student:
    def __init__(self,number, name):
        self.number = number
        self.name = name

student1 = Student(number=1, name='홍길동')
student2 = Student(number=1, name='홍길동')
student3 = Student(number=2, name='장보고')
student4 = Student(number=3, name='홍길동')

print(student1 == student2)  # BaseModel을 호출하지 않았을 때 기본적으로 ==는 주소를 비교하여 주소가 틀리면 False 반환 
# -> student1과 student2의 주소가 다르니 값이 같아도 False 반환
print(student1 == student3)
print(student3 == student4)
```

```python
class Student:
    def __init__(self,number, name):
        self.number = number
        self.name = name

    def __eq__(self,student):   # __eq__를 재정의하여 주소가 아닌 값으로 비교할 수 있도록 함
        return self.number == student.number

student1 = Student(number=1, name='홍길동')
student2 = Student(number=1, name='홍길동')
student3 = Student(number=2, name='장보고')
student4 = Student(number=3, name='홍길동')

# print(student1 == student2) 
# print(student1 == student3)
# print(student3 == student4)

print(student1.__eq__(student2)) # 값끼리 비교했기에 True 반환
print(student1 == student2) # ==이라는 연산자는 클레스 안의 __eq__연산자를 호출함 -> 클레스 안에 __eq__이 재정의되어있으므로 True 반환
# (클레스 안에 __eq__가 재정의되어있지 않다면 False)
```
