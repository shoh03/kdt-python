```python
def mean_grade(kor, eng, math, **keywards):
    mg = (kor + eng + math) / 3
    if 'num' in keywards:
        if 'name' in keywards:
            return f'{keywards['name']}님의 국,영,수 평균 : {round(mg,keywards['num'])}'
        return f'평균 : {round(mg,keywards['num'])}'
    if 'name' in keywards:
        return f'{'name'}님의 국,영,수 평균 : {round(mg,2)}'
    return f'평균 : {round(mg,2)}'
```

```python
mean_grade(80,80,90,name='오승현',num=3)
```

```python
def mean_grade(**keywards):
    kor, eng, math = keywards.get('kor',0), keywards.get('eng',0), keywards.get('math',0)
    total = kor + eng + math
    average = total / 3
    average = round(average, keywards.get('round',2))
    if 'name' in keywards:
        name = keywards['name']
        return f'{name}님의 평균은 : {average}'
    return f'평균은 : {average}'
```

```python
mean_grade(kor=100,eng=70,math=60,name = '오승현')
```
